# Threshold Tunning 

```python
from sklearn.metrics import precision_score, recall_score, f1_score
```

```python
import numpy as np

thresholds_to_test = np.arange(0.1, 0.91, 0.05)

results = []

for threshold in thresholds_to_test:
    y_pred_threshold = (y_prob_rf >= threshold).astype(int)
    
    precision = precision_score(y_test, y_pred_threshold)
    recall = recall_score(y_test, y_pred_threshold)
    f1 = f1_score(y_test, y_pred_threshold)
    
    results.append([threshold, precision, recall, f1])

threshold_results = pd.DataFrame(
    results,
    columns=['Threshold', 'Precision', 'Recall', 'F1']
)

threshold_results
```

# Threshold Performance comparison 

```python
plt.figure(figsize=(10, 6))

plt.plot(
    threshold_results['Threshold'],
    threshold_results['Precision'],
    marker='o',
    label='Precision'
)

plt.plot(
    threshold_results['Threshold'],
    threshold_results['Recall'],
    marker='o',
    label='Recall'
)

plt.plot(
    threshold_results['Threshold'],
    threshold_results['F1'],
    marker='o',
    label='F1-Score'
)

plt.xlabel('Classification Threshold')
plt.ylabel('Score')
plt.title('Precision, Recall and F1-Score vs Threshold')
plt.legend()
plt.grid()
plt.show()
```

# Final Model Conclusion 

### Conclusion

The fraud detection project was developed using Exploratory Data Analysis (EDA),
data preprocessing, class imbalance handling using SMOTE, and multiple machine
learning classification models.

Three models were evaluated:

- Logistic Regression
- Decision Tree
- Random Forest

Among these models, Random Forest performed the best overall, achieving:

- Accuracy: 99.95%
- Precision: 91.25%
- Recall: 76.84%
- F1-Score: 83.43%
- ROC-AUC: 0.964

Random Forest was selected as the final model because it provided the best
balance between precision and recall and achieved the highest F1-score.

Since fraud detection is an imbalanced classification problem, F1-score,
precision, recall and ROC-AUC were considered more important than accuracy alone.

The analysis also identified important features such as V14, V10, V12 and V17,
which contributed significantly to fraud classification.
